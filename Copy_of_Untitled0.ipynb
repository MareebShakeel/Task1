```python
%%writefile run_urdu.py
import torch
import torchaudio
from models import StyleTTS2  # Adjust if needed

device = "cuda" if torch.cuda.is_available() else "cpu"
model = StyleTTS2.load_from_checkpoint("checkpoints/styletts2.pt", map_location=device)
model.eval()

text = "السلام علیکم، آپ کیسے ہیں؟"
wav = model.infer(text)
torchaudio.save("urdu_output.wav", wav.unsqueeze(0), 24000)
```
